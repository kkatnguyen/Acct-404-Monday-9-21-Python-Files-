# Visualization using MatPlotLib and Seaborn

```python
# Total value by Type into Bar Chart 
total=data2026.groupby(['Type'])['Total Bulk and Bags'].sum()

# Plot Bar Chart
total.plot(kind='bar', figsize=(6,6))

# Print the value 
print(total)
plt.show()
```

```python
# Filter out Data
ConventionalOnly=data2026[data2026['Type']!="Organic"]
```

```python
ConventionalOnly.describe()
```

```python
# Add a revenue colummns whihc calculates ASP times Total Bulk and Bags
data2026['Revenue']=data2026['ASP Current Year']*data2026['Total Bulk and Bags']
```

```python
data2026.head()
```

```python
# Bar Chart shworing Revenue by Geography

revenue_by_geography=data2026.groupby(['Geography'])['Revenue'].sum()

# Bar Chart
revenue_by_geography.plot(kind='bar', figsize=(25,8))
```

```python
# Filter out summary data ##Multiple deletes

revenue_by_geography=data2026[data2026['Geography']!='Total U.S.']
```

```python
# Bar Chart shworing Revenue by Geography

revenue_by_geography1=revenue_by_geography.groupby(['Geography'])['Revenue'].sum()

# Bar Chart
revenue_by_geography1.plot(kind='bar', figsize=(25,8))
```

```python
# add a data sort to the chart
revenue_by_geography1_sorted=revenue_by_geography1.sort_values(ascending=False).head(63)
```

```python
# Bar Chart
revenue_by_geography1_sorted.plot(kind='bar', figsize=(14,7))
```

```python
# See distribution of Average Selling Price

# Histogram visualization to check values in data frame

plt.figure(figsize=(10,10))
numeric_values=['ASP Current Year']
data2026[numeric_values].hist(bins=20,color="purple",figsize=(20,10))
plt.show()
```

# Pie Chart

```python
# Insert Pie Chart
labels='4225','4770','4046'
segments=data2026[['4046','4225','4770']].sum()
explode=(.1,0,0)

fig1,ax1=plt.subplots()
ax1.pie(segments,explode=explode,labels=labels,shadow=True)
ax1.axis('equal') #means equal aspects or circle

plt.show()
```
